Exploratory Data Analysis (EDA)
EDA is performed to understand the characteristic of the dataset through statistical summaries and visualisations. It helps identify patterns , trends , relationships between variables and potential factors influencing customer churn.


#### Distribution of customer churn

The 'churn' column is the target variable of this project. It indicates whether a customer has left the company's service.
Analyzing the distribution of the target variable helps us understand the proportion of customers who stayed and those who churned. This also helps determine whether the dataset is balanced or imbalanced.

```python
df["Churn"].value_counts()
```

```python
df["Churn"].value_counts(normalize=True)*100
```

```python
(df["Churn"].value_counts(normalize=True)*100).round(2)#rounding up in % value
```

** Count Plot of Customer Churn
A count Plot is used to visualize the number of customers who stayed and those who churned.This provides a clear understanding of distribution of the target variable.


```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(6,4))

sns.countplot(x="Churn", data = df)

plt.title("Customer Churn Distribution")
plt.xlabel("Churn")
plt.ylabel("Number of Customers")
plt.show()
```

#### Observation

The majority of customers did not churn.

Approximately 73.42% of customers stayed with the company.

Arround 26.58% of customers churned.

The dataset is moderately imbalanced.

#### Feature Scaling

Feature scaling standardizes numerical features so they have a similar range. This helps distance-based machine learning algorithms perform better and prevents features with larger values from dominating the learning process.

In this project, StandardScaler is used to standardize the features

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

# Fit on training data and transform
x_train_scaled = scaler.fit_transform(x_train)

# Transform test data
x_test_scaled = scaler.transform(x_test)

print("Scaled Training Shape :", x_train_scaled.shape)
print("Scaled Testing Shape  :", x_test_scaled.shape)
```

#### Business Insight

#### Handling Class Imbalance

The target variable (Churn) is imbalanced, meaning there are significantly more customers who did not churn than customers who did churn.

Machine learning models trained on imbalanced data may become biased toward the majority class.

To overcome this problem, we use SMOTE (Synthetic Minority Oversampling Technique), which generates synthetic samples for the minority class, creating a more balanced dataset.

```python
y_train.value_counts()
```

```python
import sklearn
import imblearn
import threadpoolctl
import numpy as np

print("Scikit-learn :", sklearn.__version__)
print("Imbalanced-learn :", imblearn.__version__)
print("Threadpoolctl :", threadpoolctl.__version__)
print("NumPy :", np.__version__)
```

```python
import threadpoolctl
print(threadpoolctl.__version__)
```

```python
from imblearn.over_sampling import SMOTE

smote = SMOTE(random_state=42)

x_train_smote, y_train_smote = smote.fit_resample(
    x_train_scaled,
    y_train
)

print("Before SMOTE:")
print(y_train.value_counts())

print("\nAfter SMOTE:")
print(y_train_smote.value_counts())

```

#### Business Insight

Before applying SMOTE, the training dataset was imbalanced, with significantly fewer churned customers than non-churned customers (4130 vs. 1495). Such an imbalance can bias machine learning models toward predicting the majority class.

After applying SMOTE, both classes contain 4,130 samples, creating a balanced dataset. This allows the model to learn the characteristics of both churned and non-churned customers more effectively, improving its ability to identify customers who are likely to churn.

```python
from sklearn.linear_model import LogisticRegression

lr = LogisticRegression(random_state=42)

lr.fit(x_train_smote, y_train_smote)

y_pred_lr = lr.predict(x_test_scaled)
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

print("Accuracy :", accuracy_score(y_test, y_pred_lr))
print("Precision:", precision_score(y_test, y_pred_lr))
print("Recall   :", recall_score(y_test, y_pred_lr))
print("F1 Score :", f1_score(y_test, y_pred_lr))
```

A high recall is desirable for churn prediction because identifying potential churners is more important than missing them.

F1 Score = 61.33%

The model has good recall but relatively lower precision, resulting in a moderate F1-score.
